```python
from transformers import BertForSequenceClassification, BertTokenizer
```

```python
model_name = "textattack/bert-base-uncased-imdb"
```

```python
model = BertForSequenceClassification.from_pretrained(model_name, num_labels=2) #initializing the model object
```

```python
from transformers import BertTokenizer # importing tokenizer
tokenizer = BertTokenizer.from_pretrained(model_name)
tokenizer.vocab_size # vocabulary size
```

```python
inputs = tokenizer("I heart Generative AI", return_tensors = "pt") # tokenization
```

```python
! pip install torch
```

```python
import torch # making prediction with pre trained model
with torch.no_grad():
  outputs = model(**inputs).logits
  probabilities = torch.nn.functional.softmax(outputs, dim=1)
  predicted_class = torch.argmax(probabilities)
```

```python
if predicted_class == 1:
    print(f"Sentiment: Positive ({probabilities[0][1] * 100:.2f}%)")
else:
    print(f"Sentiment: Negative ({probabilities[0][0] * 100:.2f}%)")
```

```python
from transformers import DistilBertForSequenceClassification, DistilBertTokenizer, TrainingArguments, Trainer
```

```python
from datasets import load_dataset
```

```python
model = DistilBertForSequenceClassification.from_pretrained("distilbert-base-uncased", num_labels=2) # model
```

```python
tokenizer = DistilBertTokenizer.from_pretrained("distilbert-base-uncased")
```

```python
def tokenize_function(examples):
  return tokenizer(examples["text"], padding = "max_length", truncation = True) # tokenization function
```

```python
dataset = load_dataset("imdb")
```

```python
tokenized_datasets = dataset.map(tokenize_function, batched = True)
```

```python
training_args = TrainingArguments(
    per_device_train_batch_size = 64,
    output_dir = "./results",
    learning_rate = 2e-5,
    num_train_epochs = 3,
)
```

```python
trainer = Trainer(
    model = model,
    args = training_args,
    train_dataset = tokenized_datasets["train"],
    eval_dataset = tokenized_datasets["test"]
)
```

```python
trainer.train() # training the model
```
