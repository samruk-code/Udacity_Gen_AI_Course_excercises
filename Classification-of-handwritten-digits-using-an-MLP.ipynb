### Importing libraries 

```python
import matplotlib.pyplot as plt 
import numpy as np
import tensorflow as tf
```

### Get the data from Tensorflow Datasets 

```python
mnist = tf.keras.datasets.mnist
(training_images, training_labels), (test_images, test_labels) = mnist.load_data()
```

### Lokking the shape of the features and labels

```python
print(f"Training Images Shape: {training_images.shape}")
print(f"Training Labels Shape: {training_labels.shape}")
print(f"Test Images Shape: {test_images.shape}")
print(f"Test Labels Shape: {test_labels.shape}")
```

### Normalizing data for fast training

```python
training_images = training_images / 255.0
test_images = test_images / 255.0
```

```python
# taking a look at a training data 
print("Corresponding label for following Image: ", training_labels[:1])
plt.imshow(training_images[:1].reshape(28,28), cmap = plt.cm.gray)
```

```python
# Multilayer Perceptron for classifying handwritten digit 
from sklearn.neural_network import MLPClassifier

mlp = MLPClassifier(
    hidden_layer_sizes = (50, ),
    max_iter = 10,
    alpha = 1e-4,
    solver = "sgd",
    verbose = 10,
    random_state = 1,
    learning_rate_init = 0.1,
)
```

```python
# fit the data with training data and labels
mlp.fit(training_images.reshape(60000, -1), training_labels)
```

```python
# Calculating Training and Test Accuracy
print(mlp.score(training_images.reshape(60000, -1), training_labels))
print(mlp.score(test_images.reshape(10000, -1), test_labels))
```

```python
# Making Prediction 
print("True_label : ", test_labels[:1])
mlp.predict(test_images[:1].reshape(1,-1))
```
