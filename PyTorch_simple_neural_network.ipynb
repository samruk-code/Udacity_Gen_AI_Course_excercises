### Simple PyTorch Project from building dataset, architecting the network and training and prediction 

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
```

```python
class NumberSumDataset(Dataset):
    def __init__(self, data_range = (1, 10)):
        self.numbers = list(range(data_range[0], data_range[1]))

    def __getitem__(self, index):
        number1 = float(self.numbers[index // len(self.numbers)])
        number2 = float(self.numbers[index % len(self.numbers)])
        return torch.tensor([number1, number2]), torch.tensor([number1 + number2])

    def __len__(self):
        return len(self.numbers) ** 2
```

```python
dataset = NumberSumDataset(data_range = (1, 100))
```

```python
class MPL(nn.Module):
    def __init__(self, input_size):
        super(MPL, self).__init__()
        self.hidden_layer = nn.Linear(input_size, 128)
        self.output_layer = nn.Linear(128, 1)
        self.activation = nn.ReLU()

    def forward(self, x):
        x = self.activation(self.hidden_layer(x))
        return self.output_layer(x)
```

```python
dataset = NumberSumDataset(data_range = (1, 100))
dataloader = DataLoader(dataset, batch_size = 100, shuffle = True)
model = MPL(input_size = 2)
loss_function = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr = 0.001)
```

```python
for epoch in range(10):
    total_loss = 0.0 
    for number_pair, sums in dataloader:
        predictions = model(number_pair)
        loss = loss_function(predictions, sums)
        loss.backward()
        optimizer.step()
        optimizer.zero_grad()

        total_loss = total_loss + loss.item()
    print("Epoch {}: Sum of Batch Losses = {:.5f}".format(epoch+1, total_loss))
        
```
